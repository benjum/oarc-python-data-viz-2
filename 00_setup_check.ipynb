Run every cell in this notebook. If each cell prints ✅ and shows a chart, you're ready. If something fails, see the troubleshooting notes at the bottom.

````python
import importlib, sys

print("Python", sys.version.split()[0])
packages = ["pandas", "numpy", "matplotlib", "plotly", "altair", "streamlit",
            "dash", "holoviews", "hvplot", "panel", "datashader", "bokeh"]
for name in packages:
    try:
        mod = importlib.import_module(name)
        print(f"✅ {name:<11} {mod.__version__}")
    except Exception as e:
        print(f"❌ {name:<11} {type(e).__name__}: {e}")
````

You should see an interactive Plotly chart below (hover over a dot):

````python
import plotly.express as px

px.scatter(px.data.iris(), x="sepal_width", y="sepal_length", color="species",
           title="✅ Plotly renders")
````

And an Altair chart (drag inside it to zoom and pan):

````python
import altair as alt

alt.Chart(px.data.iris()).mark_point().encode(
    x="sepal_width", y="sepal_length", color="species"
).properties(title="✅ Altair renders").interactive()
````

And a HoloViews/Bokeh chart via hvPlot:

````python
import hvplot.pandas  # adds .hvplot to DataFrames

px.data.iris().hvplot.scatter(x="sepal_width", y="sepal_length", by="species",
                              title="✅ hvPlot renders")
````

### Troubleshooting

| Symptom | Fix |
| --- | --- |
| `ModuleNotFoundError` | `pip install -r requirements.txt` from the workshop folder, then restart the kernel |
| Plotly chart is blank in VS Code / classic Notebook | `import plotly.io as pio; pio.renderers.default = "notebook"` (or `"vscode"`) |
| Altair chart is blank or shows JSON | Use JupyterLab ≥ 4, or run `alt.renderers.enable("jupyter", offline=True)` (needs `anywidget`) |
| hvPlot chart doesn't appear | Run `import holoviews as hv; hv.extension("bokeh")` and re-run the cell |
| `streamlit: command not found` | Run `python -m streamlit hello` instead |
